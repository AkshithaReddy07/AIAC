```python
def is_leap_year(year):
  """
  Checks if a given year is a leap year.

  Args:
    year: The year to check (integer).

  Returns:
    True if the year is a leap year, False otherwise.
  """
  return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)

# Example usage:
year_to_check = 2024
if is_leap_year(year_to_check):
  print(f"{year_to_check} is a leap year.")
else:
  print(f"{year_to_check} is not a leap year.")

year_to_check = 1900
if is_leap_year(year_to_check):
  print(f"{year_to_check} is a leap year.")
else:
  print(f"{year_to_check} is not a leap year.")

year_to_check = 2000
if is_leap_year(year_to_check):
  print(f"{year_to_check} is a leap year.")
else:
  print(f"{year_to_check} is not a leap year.")
```

```python
def cm_to_inches(cm):
  """
  Converts centimeters to inches.

  Args:
    cm: The length in centimeters (float or integer).

  Returns:
    The length in inches (float).
  """
  return cm * 0.393701

# Example usage:
centimeters = 10
inches = cm_to_inches(centimeters)
print(f"{centimeters} centimeters is equal to {inches} inches.")

centimeters = 25.4
inches = cm_to_inches(centimeters)
print(f"{centimeters} centimeters is equal to {inches} inches.")
```

```python
def format_full_name():
  """
  Takes a full name as input and formats it into first and last names.
  """
  full_name = input("Please enter your full name: ")

  # Split the full name into parts
  name_parts = full_name.split()

  if len(name_parts) >= 2:
    first_name = name_parts[0]
    last_name = name_parts[-1]
    print(f"First Name: {first_name}")
    print(f"Last Name: {last_name}")
  elif len(name_parts) == 1:
    print(f"First Name: {name_parts[0]}")
    print("Please provide a last name as well.")
  else:
    print("Invalid input. Please enter your full name.")

# Run the program
format_full_name()
```

```python
def count_vowels(text):
  """
  Counts the number of vowels in a given string.

  Args:
    text: The input string.

  Returns:
    The number of vowels in the string.
  """
  vowels = "aeiouAEIOU"
  vowel_count = 0
  for char in text:
    if char in vowels:
      vowel_count += 1
  return vowel_count

# Get input from the user
input_string = input("Enter a string: ")

# Count and print the vowels
num_vowels = count_vowels(input_string)
print(f"The number of vowels in '{input_string}' is: {num_vowels}")

# Example demonstrating with your input
example_string = "Akshitha Reddy"
example_vowels = count_vowels(example_string)
print(f"The number of vowels in '{example_string}' is: {example_vowels}")
```

```python
def count_lines_in_file(filepath):
  """
  Reads a file and counts the number of lines.

  Args:
    filepath: The path to the file (string).

  Returns:
    The number of lines in the file (integer), or -1 if the file cannot be read.
  """
  try:
    with open(filepath, 'r') as file:
      line_count = sum(1 for line in file)
    return line_count
  except FileNotFoundError:
    print(f"Error: File not found at {filepath}")
    return -1
  except Exception as e:
    print(f"An error occurred: {e}")
    return -1

# Example usage:
# Create a dummy file for demonstration
with open("my_example_file.txt", "w") as f:
  f.write("This is line 1.\n")
  f.write("This is line 2.\n")
  f.write("This is line 3.\n")

file_path = "my_example_file.txt"
line_count = count_lines_in_file(file_path)

if line_count != -1:
  print(f"The file '{file_path}' has {line_count} lines.")

# Example with a non-existent file
non_existent_file = "non_existent_file.txt"
count_lines_in_file(non_existent_file)
```
